THIS CODE WILL DETECT ONLY PERSONS IN THE VIDEO BECAUSE WE ARE SUING (CLASSES = 0)

```python
from ultralytics import YOLO 

#Load a Model ----------------------------> 
model = YOLO('yolo11m.pt')

#Perform object detection on video ------------------> 
result = model("/kaggle/input/datasets/harshal63234/detection-video/mixkit-people-walking-in-the-street-in-japan-4437-hd-ready.mp4",classes=0,save=True)
```

NOW TRACK THE PERSONS USING BOTSORT ALGORITHM ------------------> 

```python
result = model.track(source="/kaggle/input/datasets/harshal63234/detection-video/mixkit-people-walking-in-the-street-in-japan-4437-hd-ready.mp4",persist=True,classes=0,save=True)
```

IF YOU WANT TO DO TRACKIG USING THE (bytetrack algorithm) then the code will like this ------------->

```python
result = model.track(source="/kaggle/input/datasets/harshal63234/detection-video/mixkit-people-walking-in-the-street-in-japan-4437-hd-ready.mp4",persist=True,classes=0,save=True,tracker="bytetrack.yaml")
```
