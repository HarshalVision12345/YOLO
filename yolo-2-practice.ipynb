```python
from ultralytics import YOLO 

# Load a COCO-pretraind YOLO11s-cls model 
model = YOLO("yolo11s-cls.pt")

#Train The Model 
result = model.train(data="/kaggle/input/datasets/harshal63234/custom-datatset/custom_data",epochs=10,imgsz=640)
```

Run inference with the custom classification model 

```python
# Run inference with the YOLO11s model on the image 
model = YOLO ("/kaggle/working/runs/classify/train/weights/best.pt")

result = model("/kaggle/input/datasets/harshal63234/test-data/test/dandelion/dandelion_test_02.jpg")
result[0].show()
```
